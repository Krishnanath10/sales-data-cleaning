```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("Products.csv")
df1.head(10)
```

```python
# clean whitespace and formatting the headeres

df1.columns = df1.columns.str.lower().str.strip().str.replace(" ", "_")
df1.columns
```

```python
# remove unwanted characters from column values

colms = ["product_name", "brand"]

for i in colms:
    df1[i] = (
        df1[i].str.replace(r"^[A-Z]\.\s*","", regex=True).str.strip()
    )

    
```

```python
df1.iloc[1000:1030]
```

```python
# remove $ from price related columns

mony_col = ["unit_cost_usd", "unit_price_usd"]

for i in mony_col:
    df1[i] = (
        df1[i].str.replace("$","", regex=False).str.replace(",","", regex=False)
    )
    df1[i]= pd.to_numeric(df1[i], errors="coerce")
```

```python
df1.iloc[1950:1960]
```

```python
# saved csv file in local folder

df1.to_csv("updated_data/Product1.csv", index=False)
```

```python
import pandas as pd
import numpy as np
```

```python
df_sal = pd.read_csv("Sales.csv")
df_sal
```

```python
# update the column names and remove space and replace it with "_", it will be easy while using this data into SQL

df_sal.columns = df_sal.columns.str.lower().str.strip().str.replace(" ", "_")
```

```python
# change the order_date and delivery_date format is not correct, so update the format to datetime

date_col = ["order_date", "delivery_date"]

for i in date_col:
    df_sal[i] = (
        df_sal[i].str.strip()
    )
    df_sal[i] = pd.to_datetime(df_sal[i], errors="coerce")
```

```python
df_sal.head(10)
```

```python
# checking the null values

print("Invalid Dates:", df_sal["delivery_date"].isna().sum())
```

```python
# checking total number of rows we have in data also how many rows of delivery_date column is having empty cell

print("Total rows:", len(df_sal))
print("Missing delivery date:", df_sal["delivery_date"].isna().sum())
```

```python
# hear we are checking % of missing values of the total data, and found 79.06% of the data is not having the any value in delivery_date column

missing_perc = df_sal["delivery_date"].isna().mean() * 100
print("missinge per%: ", round(missing_perc, 2))
```

```python
# we are checking number of orders placed by the order_number and how many of those order numbers are not having the delivery date

order_check = (
    df_sal.groupby("order_number")
    .agg(
        total_lines = ("line_item", "count"),
        missing_delivery = ("delivery_date", lambda x: x.isna().sum())
    )
)

print(order_check.head(20))
```

```python
# like in previous code we found that there are some order which has the delivery dates, for ex. if one order number is having 3 products
# so out of those 3, 2 products are having the delivery date, so here we are checking how many order_number are not having the delivery date

fully_missing_orders = (
    order_check["total_lines"] == order_check["missing_delivery"]
).sum()

print("Order with completely missing with delivery date: ", fully_missing_orders)
```

```python
# here we are checking how many orders are having the delivery date

orders_with_delivery= (
    order_check["missing_delivery"] < order_check["total_lines"]
).sum()

print("orders with delivery date: ", orders_with_delivery)
```

```python
# in thise we are getting same information, but in different way

order_check["delivery_status"] = "Partial"

order_check.loc[
    order_check["missing_delivery"] == order_check["total_lines"],
    "delivery_status"
] = "All Missing"

order_check.loc[
    order_check["missing_delivery"] == 0,
    "delivery_status"
] = "All available"

print(order_check["delivery_status"].value_counts())
```

```python
# here we are checking the date range of missing delivery dates and available deliery date

order_check = order_check.reset_index()

order_with_date = df_sal.merge(
    order_check[["order_number", "delivery_status"]],
    on= "order_number",
    how= "left"
)

print(
    order_with_date.groupby("delivery_status")["order_date"]
    .agg(["min", "max", "count"])
)
```

```python
# here we are checking in which year how many orders delivered and how many not delivered

print(
    order_with_date
    .groupby(
        [order_with_date["order_date"].dt.year, "delivery_status"]
    )
    .size()
)
```

```python
# we are checking the same information in %

year_delivery = pd.crosstab(
    order_with_date["order_date"].dt.year,
    order_with_date["delivery_status"],
    normalize= "index"
) * 100

print(year_delivery.round(2))
```

```python
# we have created another column which give me the True and False status of delivery date status, if order_number is not having the delivery date
# it will give me False and if has then it will give me True
df_sal["delivery_date_available"] = (
    df_sal["delivery_date"].notna()
)
```

```python
# same information getting as available and missing in different column

df_sal["delivery_date_status"] = df_sal["delivery_date"].notna().map({
    True: "Available",
    False: "Missing"
})
```
